```
# 2. Initialize and run CausalNerve on NASA C-MAPSS FD001
# The preset automatically downloads the dataset and configures the 21-sensor graph
from causalnerve import CausalNerve
from causalnerve.datasets import CMAPSSDataset

print('Downloading and loading NASA C-MAPSS FD001...')
dataset = CMAPSSDataset(subset='FD001', include_settings=False)
engine_data = dataset.load_engine(1).X
nerve = CausalNerve.from_preset('turbofan')

print('Fitting baseline causal structure...')
nerve.fit(engine_data[:50], epochs=20)

print('Monitoring for structural degradation...')
nerve.watch(engine_data[50:])
```

### What just happened?
CausalNerve established a baseline causal graph from healthy engine data, then actively monitored the streaming test telemetry. When the engine began degrading, the system autonomously detected structural divergence (increased thermodynamic entropy) and performed causal surgery to adapt its graph to the new failure mode.

```
# 3. Root Cause Analysis
# We ask the engine to explain the causal origin of the Exhaust Gas Temperature (EGT) alarm.
print('\nRoot Cause Analysis for EGT:')
nerve.why('S3')
```

```
# 4. Counterfactual Simulation
# What would happen to the rest of the system if we manually intervened to stabilize the High-Pressure Turbine (HPT)?
print('\nSimulating Intervention (do-calculus) on S3 (EGT):')
nerve.what_if('S3', 0.0)
```
