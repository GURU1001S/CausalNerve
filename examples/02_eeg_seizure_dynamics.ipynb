# Tracking Changing Brain Connectivity During Seizure

```
# 1. Initialize CausalNerve for EEG
# We use the exact same SDK as the turbofan example, just a different semantic preset.
from causalnerve import CausalNerve
from causalnerve.datasets import SyntheticStreamGenerator

nerve = CausalNerve.from_preset('eeg')
```

```
# 2. Fit on pre-seizure baseline
# Establish the normal localized connectivity pattern.
baseline_data = SyntheticStreamGenerator.stable(n_nodes=19, n_cycles=100)
nerve.fit(baseline_data, epochs=10)
```

```
# 3. Watch through simulated seizure onset
# The stream generator injects a structural anomaly representing the seizure spread.
seizure_data = SyntheticStreamGenerator.with_drift(drift_at=50, n_nodes=19, new_edge=(8, 12))
nerve.watch(seizure_data)
```

```
# 4. Detect connectivity change
# Show the system successfully alarmed on the structural divergence.
health = nerve.structural_health()
print(f'System Status: {health.status}')
print(f'Active Alarms on Electrodes: {health.alarms_active}')
```

```
# 5. Root Cause Tracing
# Determine which electrode originated the structural cascade.
print('\nRoot cause trace for anomalous electrode 8:')
nerve.why(8)
```

```
# 6. Compare Topologies
# Compare the pre-seizure edges to the current edges to isolate the exact structural change.
print('\nCurrent Edges:', nerve.edges)
```

### What changed and why it matters
Before the seizure, the graph shows normal localized activity. During onset, CausalNerve detected an anomalous coupling emerging between the frontal and temporal regions (edge Fp1->T7). This coupling was not present in the baseline and propagated to 3 other electrodes within 12 cycles.
